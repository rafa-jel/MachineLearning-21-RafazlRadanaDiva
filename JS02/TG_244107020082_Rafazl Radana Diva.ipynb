PyPREP: Pustaka yang digunakan khusus untuk prapemrosesan (preprocessing) data sinyal EEG (rekam aktivitas gelombang otak). Fungsinya mencakup pembersihan data secara otomatis, seperti mendeteksi dan mengabaikan saluran (channel) sensor yang buruk, serta melakukan standardisasi referensi sinyal agar data siap dianalisis.

Scipy: Pustaka fundamental untuk komputasi ilmiah dan matematis di Python (dibangun di atas Numpy). Library ini menyediakan berbagai modul untuk perhitungan teknis tingkat lanjut, seperti optimasi algoritma, aljabar linear, kalkulus (integral dan diferensial), interpolasi, serta pemrosesan sinyal dan gambar.

wandb (Weights & Biases): Platform yang digunakan dalam Machine Learning dan Deep Learning untuk melacak (tracking) jalannya eksperimen. Kegunaannya adalah memantau kinerja model secara real-time (seperti tingkat akurasi dan persentase error), menyimpan log parameter pelatihan, serta membuat visualisasi dan laporan untuk membandingkan mana model AI yang paling optimal.

pyECG: Pustaka yang didesain secara spesifik untuk memproses dan menganalisis sinyal ECG/EKG (rekam kelistrikan jantung). Kegunaan utamanya meliputi ekstraksi fitur dari sinyal jantung, pendeteksian gelombang kompleks QRS (detak jantung utama), serta komputasi untuk analisis Heart Rate Variability (HRV).
